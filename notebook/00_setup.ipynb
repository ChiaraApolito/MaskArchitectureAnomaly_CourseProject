# 00 — Colab environment setup

**Run this notebook FIRST** in every Colab session. It does three things:
1. mounts Google Drive;
2. clones the repo onto Drive (or **forces** it to the latest version of the branch);
3. installs the pip dependencies (`eomt/requirements.txt` + `ood_metrics`, `xlsxwriter`).

Then **close this notebook** and open `Task4` / `Task5` / `Task7` / `Task8` on the **same Colab server**: they will find Drive mounted, the repo cloned, and the packages installed.

```python
from google.colab import drive
drive.mount("/content/drive")

import os, sys, subprocess, shutil

REPO = "/content/drive/MyDrive/MaskArchitectureAnomaly_CourseProject"
BRANCH = "finetuning/coco-to-cityscapes"
URL = "https://github.com/ChiaraApolito/MaskArchitectureAnomaly_CourseProject.git"

# Move outside the repo: if the cwd is inside REPO (e.g. a Task already did chdir),
# emptying it would prevent git from reading the working directory.
os.chdir("/content")

# 1) Clean clone. On Google Drive (FUSE) rmtree often does NOT remove the folder,
#    so we empty its contents by hand: git can clone into an existing but empty dir.
if os.path.exists(REPO):
    print("Emptying the existing copy...")
    for name in os.listdir(REPO):
        p = os.path.join(REPO, name)
        shutil.rmtree(p, ignore_errors=True) if os.path.isdir(p) else os.remove(p)

print("Cloning the repo...")
r = subprocess.run(
    ["git", "clone", "--branch", BRANCH, "--single-branch", URL, REPO],
    capture_output=True, text=True, cwd="/content",
)
if r.returncode != 0:
    print(r.stdout)
    print(r.stderr)
    raise RuntimeError(f"git clone failed (exit {r.returncode})")

# 2) Install dependencies (EoMT requirements + extras used by the tasks).
print("Installing dependencies...")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", REPO + "/eomt/requirements.txt"])

print("\nSetup complete. Open Task4 / Task5 / Task7 / Task8 on the same Colab server.")
```
