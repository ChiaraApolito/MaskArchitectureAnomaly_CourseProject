# 00 — Setup ambiente Colab

**Esegui questo notebook per PRIMO** in ogni sessione Colab. Fa tre cose:
1. monta Google Drive;
2. clona la repo su Drive (o la **forza** all'ultima versione del branch);
3. installa le dipendenze pip (`eomt/requirements.txt` + `ood_metrics`, `xlsxwriter`).

Poi **chiudi questo notebook** e apri `Task4` / `Task5` / `Task7` / `Task8` nello **stesso server Colab**: troveranno Drive montato, repo clonata e pacchetti installati.

```python
from google.colab import drive
drive.mount("/content/drive")

import os, sys, subprocess, shutil

REPO = "/content/drive/MyDrive/MaskArchitectureAnomaly_CourseProject"
BRANCH = "finetuning/coco-to-cityscapes"
URL = "https://github.com/ChiaraApolito/MaskArchitectureAnomaly_CourseProject.git"

# 1) Clonazione "pulita": cancella la cartella e ri-clona da zero, cosi' la repo e'
#    sempre identica al branch remoto (nessuna modifica locale / stato stantio).
if os.path.exists(REPO):
    print("Rimuovo la copia esistente...")
    shutil.rmtree(REPO, ignore_errors=True)

print("Clono la repo...")
subprocess.run(["git", "clone", "--branch", BRANCH, "--single-branch", URL, REPO], check=True)

# 2) Installa le dipendenze (requirements EoMT + extra usati dai task).
print("Installo le dipendenze...")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", REPO + "/eomt/requirements.txt"])
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "ood_metrics", "xlsxwriter"])

print("\nSetup completato. Apri Task4 / Task5 / Task7 / Task8 nello stesso server Colab.")
```
